```python
#1
import pandas as pd


df = pd.read_excel("catalog_products.xlsx")
print('сколько всего строк и столбцов:' , df.shape)
print('какие типы данных: ',df.dtypes)
print('пропущенные значения:', df.isnull().sum())
print(df.head())
```

```python
#2
for c in df.columns:
    con=pd.to_numeric(df[c], errors='coerce')
    if con.notnull().sum()>(len(df)/2):
        df[c]=con.fillna(con.mean()).astype(float)
    else:
        pass

n=df.select_dtypes(include=['number']).columns

print(df.head())
```

```python
#3
import numpy as np
df['total_value']=df['col_2']*df['col_3']
df['double_stock']=df['col_3']*2
df['log_price']=np.log(df['col_2'])
print(df.head())
```

```python
#4
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(20, 15))
plt.subplot(2, 2, 1)
sns.histplot(df['col_2'], color='green')
plt.subplot(2, 2, 2)
sns.scatterplot(data=df, x='col_2', y='col_3')
plt.subplot(2, 1, 2)
sns.boxplot(data=df, x='col_2', y ='col_7', hue='col_7')
plt.show()
```

```python
#5
m=df['col_2'].mean()
s=df['col_2'].std()
more=m+3*s
l=m-3*s
anomalies=df[(df['col_2']>more)|(df['col_2']<l)]
clean=df[(df['col_2']<=more)&(df['col_2']>=l)]
print(anomalies)
print(clean)
```

```python
#6
df_num=pd.get_dummies(df,columns=['col_7'],drop_first=True)
print(df_num.dtypes.head(10))
print(df_num.head())
```

```python
#7
from sklearn.model_selection import train_test_split
y=df_num['col_2']
x=df_num.drop('col_2',axis=1)
x=x.select_dtypes(include=['int64','float64','uint8','bool'])
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#8
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error
model=LinearRegression()
model.fit(x_train,y_train)
y_pred =model.predict(x_test)
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
print(mae)
print(mse)
```

```python
#9
df_feat=df_num.copy()
df_feat['total_value']=df_feat['col_2']*df_feat['col_3']
df_feat['double_stock']=df_feat['col_3']*2
y_m=df_feat['col_2']
x_m=df_feat.drop('col_2',axis=1)
x_m=x_m.select_dtypes(include=['int64','float64','uint8','bool'])
if 'log_price' in x_m.columns:
    x_multi = x_m.drop(['log_price'], axis=1)
x_train_m,x_test_m,y_train_m,y_test_m=train_test_split(x_m,y_m,test_size=0.2,random_state=42)
model_m=LinearRegression()
model_m.fit(x_train_m,y_train_m)
y_pred_m=model_m.predict(x_test_m)
mae_m=mean_absolute_error(y_test_m,y_pred_m)
mse_m=mean_squared_error(y_test_m,y_pred_m)
print(mae_m)
print(mse_m)
```
